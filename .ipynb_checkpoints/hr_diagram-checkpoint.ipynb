```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("./../course_assignments/homework7/star_data.txt", skiprows=1)
print(np.shape(data)) # 2642 rows
magnitude = data[:, 0]
temperature = data[:, 1]
print(min(magnitude)) # correspond to brightest star
print(max(magnitude)) # correspond to dimmest star
print(min(temperature)) # correspond to cooler stars
print(max(temperature)) # correspond to hotter stars
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
brightest_star = min(magnitude)

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.show()
```
