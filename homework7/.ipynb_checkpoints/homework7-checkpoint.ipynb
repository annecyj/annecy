```python
import numpy as np
import matplotlib.pyplot as plt
```

MatPlotLib

```python
def make_sine_wave(x, A, w):
    y = A * np.sin (w * x)
    return y
```

```python
x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(1, 3, figsize=(25, 5))

for amp, freq in zip(amplitude, frequency):
    label = f"amp={amp}"
    y = make_sine_wave(x, amp, 1)
    ax[0].plot(x, y, label = label)
ax[0].set_title("varying amplitude")
ax[0].set_xlabel("x", fontsize = 15)
ax[0].set_ylabel("y", fontsize = 15)
ax[0].legend()
ax[0].grid()

for amp, freq in zip(amplitude, frequency):
    label = f"freq={freq}"
    y = make_sine_wave(x, 1, freq)
    ax[1].plot(x, y, label = label)
ax[1].set_title("varying frequency")
ax[1].set_xlabel("x", fontsize = 15)
ax[1].set_ylabel("y", fontsize = 15)
ax[1].legend()
ax[1].grid()

for amp, freq in zip(amplitude, frequency):
    label = f"amp={amp}, freq={freq}"
    y = make_sine_wave(x, amp, freq)
    ax[2].plot(x, y, label = label)
ax[2].set_title("varying both")
ax[2].set_xlabel("x", fontsize = 15)
ax[2].set_ylabel("y", fontsize = 15)
ax[2].legend()
ax[2].grid()

plt.show()
```
