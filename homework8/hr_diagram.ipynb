```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("./../course_assignments/homework7/star_data.txt", skiprows=1)
print(np.shape(data)) # 2642 rows
magnitude = data[:, 0]
temperature = data[:, 1]
print(min(magnitude)) # corresponds to the brightest star
print(max(magnitude)) # corresponds to the dimmest star
print(min(temperature)) # corresponds to the coolest star
print(max(temperature)) # correspond to the hottest star
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(log_temperature, log_magnitude, label = "general stars", cmap = "magma", c = log_temperature)
plt.scatter(8.23275681, 0.8064758658669485, color = "black", marker='*', s=300, label = "brightest star")
plt.scatter(8.30676739, 2.5249283241374862, color = "purple", marker='*', s=300, label = "dimmest star")
plt.scatter(8.203361419071118, 2.3997118, color = "orange", marker='*', s=300, label = "coldest star")
plt.scatter(9.221555305852075, 1.52388002, color = "green", marker='*', s=300, label = "hottest star")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

cbar = plt.colorbar(scatter)
cbar.set_label("Temperature (K)")
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()
plt.show()
```

```python
point1 = np.where(log_magnitude == min(log_magnitude))
print(log_temperature[point1])
point2 = np.where(log_magnitude == max(log_magnitude))
print(log_temperature[point2])
point3 = np.where(log_temperature == min(log_temperature))
print(log_magnitude[point3])
point4 = np.where(log_temperature == max(log_temperature))
print(log_magnitude[point4])
print(min(log_magnitude))
print(max(log_magnitude))
print(min(log_temperature))
print(max(log_temperature))
```
