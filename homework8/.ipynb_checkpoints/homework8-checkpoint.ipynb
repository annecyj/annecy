```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv(r"\Users\annec\Python_DeCal_Fa2025\course_assignments\homework8\state_data.csv")
df = pd.DataFrame(data)
```

```python
print(df.shape) # 4 columns, 52 rows
```

```python
print(df.head())
print(df.tail())
# state represents us states by abbreviations, latitude represents coordinates north to south, longitude represents coordinates east to west
# Alaska
# Wyoming
# 51
# Washington DC is included
```

```python
print(df["latitude"].median()) # 39.4355155
print(df["longitude"].median()) # -89.093198
```

```python
distance_from_median_lat = df["latitude"].values - df["latitude"].median()
absolute_distance_from_median_lat = abs(distance_from_median_lat)
df["distance from median latitude"] = absolute_distance_from_median_lat
df
```

```python
distance_from_median_long = df["longitude"].values - df["longitude"].median()
absolute_distance_from_median_long = abs(distance_from_median_long)
df["distance from median longitude"] = absolute_distance_from_median_long
df
```

```python
print(df.shape) # 6 columns, 52 rows
```

```python
closest_median_lat_distance = df["distance from median latitude"].min()
# df["distance from median latitude"].idxmin()
print(f"The closest state to the median latitude is {df["name"][5]} by {closest_median_lat_distance:.2f} degrees")

farthest_median_lat_distance = df["distance from median latitude"].max()
# df["distance from median latitude"].idxmax()
print(f"The farthest state from the median latitude is {df["name"][0]} by {farthest_median_lat_distance:.2f} degrees")

closest_median_long_distance = df["distance from median longitude"].min()
# df["distance from median longitude"].idxmin()
print(f"The closest state to the median longitude is {df["name"][14]} by {closest_median_long_distance:.2f} degrees")

farthest_median_long_distance = df["distance from median longitude"].max()
# df["distance from median longitude"].idxmax()
print(f"The farthest state to the median longitude is {df["name"][11]} by {farthest_median_long_distance:.2f} degrees")
```

```python
penguin_data = sns.load_dataset("penguins")
penguins = pd.DataFrame(penguin_data)
penguins
```

```python
new_df = penguins.dropna()
new_df
```

```python
new_df.shape # 333 x 7
# 333 rows, 7 columns
```

```python
new_df.head()
```

```python
new_df.tail()
```

```python
# species -> what penguin species the penguin is
# island -> what island the penguin lives on
# bill_length_mm -> how long the bill is in mm
# bill_depth_mm -> depth of the bill in mm
# flipper_length_mm -> the length of the flipper in mm
# body_mass_g -> the mass of the penguin's body in grams
# sex -> penguin's sex
```

```python
fig, ax = plt.subplots(2, 2)

ax[0,0].scatter(new_df["bill_length_mm"], new_df["bill_depth_mm"], color = "green", label = "data points")
ax[0,0].set_title("bill length (mm) vs bill depth (mm)", fontsize = 9)
ax[0,0].set_xlabel("bill length (mm)")
ax[0,0].set_ylabel("bill depth (mm)")
ax[0,0].legend()

ax[0,1].scatter(new_df["flipper_length_mm"],	new_df["body_mass_g"], color = "purple")
ax[0,1].set_title("flipper length (mm) vs body mass (g)", fontsize = 9)
ax[0,1].set_xlabel("flipper length (mm)")
ax[0,1].set_ylabel("body mass (g)")

sns.histplot(new_df["sex"], ax=ax[1,0], shrink = 0.5, color = "blue")
ax[1,0].set_title("sex histogram", fontsize = 9)

sns.boxplot(new_df["body_mass_g"], ax=ax[1,1], color = "orange")
ax[1,1].set_title("body mass boxplot", fontsize = 9)
ax[1,1].set_ylabel("body mass (g)")

plt.tight_layout()
plt.show()

# subplot 1 compares the bill length and depth. i decided in these columns because I wanted to see if there was a correleation; if the bill length is longer, would it have more depth or less depth because it'd be too large to support?
# subplot 2 compares the flipper length and body mass. i decided on these columns because i wanted to see if longer flipper lengths meant larger body masses
# subplot 3 compares the amount of penguins by sex. i decided on this column because i wanted to see whether they got more data of female or male penguins
# subplot 4 represents the body mass in a boxplot. i decided on this column because i wanted to see if there were more massive or less massive penguins

# subplot 1 showed me there's not much of a correlation between bill length and depth. Just because it's longer, doesn't mean the depth is much greater. There is more of a chance of the bills being shorter.
# subplot 2 showed me the longer the flipper length, the more their mass is
# subplot 3 showed me the amount of penguins was about the same for each sex with only a few more males
# subplot 4 showed me there are more less massive penguins, with the median around 4000g
```
