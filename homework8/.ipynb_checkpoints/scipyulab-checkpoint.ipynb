```python
import scipy as sp
import numpy as np
from scipy import constants, integrate, optimize, interpolate
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
print("pi:", constants.pi)
print("speed of light:", constants.c, "m/s")
```

```python
# integration

def sin2(x):
    return np.sin(x)**2

# integrate from 0 to pi
result, error = integrate.quad(sin2, 0, np.pi)
print(f"integral of sin^2(x) from 0 to pi: {result}")
```

```python
# integrate x^2*e^(-x) from 0 to infinity

def new_func(x):
    return (x**2)*np.exp(-1*x)

result, error = integrate.quad(new_func, 0, np.inf)
print(result)
```

```python
# optimization

def f(x):
    return (x-3)**2 + 2

res = optimize.minimize(f, x0 = 0)
print(res)
```

```python
def fun_func(x):
    return x**4 - 3*x**3 + 2

opt = optimize.minimize(fun_func, x0 = 2)
print(opt)
```

```python
func_val = opt["fun"]
x_val = opt["x"]

print(f"min val: {func_val}, x val at min. point: {x_val}")
```

```python
# interpolating
x = np.linspace(0, 10, 10)
y = np.sin(x)

f_interp = interpolate.interp1d(x, y)

x_new = np.linspace(0, 10, 100)
y_new = f_interp(x_new)

plt.figure()
plt.scatter(x, y, label = "original values")
plt.plot(x_new, y_new, label = "interpolated values")
plt.legend()
plt.show()
```

```python
new_interp = interpolate.interp1d(x, y, kind = "cubic")

y_mod = new_interp(x_new)

plt.figure()
plt.scatter(x, y, label = "og")
plt.plot(x_new, y_mod, label = "interp")
plt.legend()
plt.show()
```

```python
# curve fitting

np.random.seed(0) # generating random data set
x = np.linspace(0, 10, 20)
y = 3.5*x + 2.0 + np.random.normal(0, 2, size=len(x))

def linear(x, m, b):
    return m*x + b

popt, pcav = curve_fit(linear, x, y)
m_fit, b_fit = popt

plt.figure()
plt.scatter(x, y, label = "nosiy data")
plt.plot(x, linear(x, *popt), color = "red", label = "fitted model")
plt.legend()
plt.show()

print(m_fit) # fitted slope
print(b_fit) # fitted intercept
```

```python
def exp_decay(x, a, b, c):
    return a * np.exp(-b * x) + c

x = np.linspace(0, 4, 50)
y = exp_decay(x, 2.5, 1.3, 0.5) + 0.2 * np.random.normal(size=len(x))

popt, _ = curve_fit(exp_decay, x, y)
plt.figure()
plt.scatter(x, y, label = "noisy data")
plt.plot(x, exp_decay(x, *popt), color = "red", label = "exp. decay fit")
plt.legend()
plt.show()
```

```python
x_data = np.linspace(-3, 3, 20)
y_data = 2*x_data**2 - 3*x_data + 1 + np.random.normal(0, 2, size=len(x_data))

def quad(x, a, b, c):
    return a*x**2 + b*x + c

popt, _ = curve_fit(quad, x_data, y_data)
plt.figure()
plt.scatter(x_data, y_data)
plt.plot(x_data, quad(x_data, popt[0], popt[1], popt[2]))
plt.show()
```
