```python
columns = ['Mass (10^24kg)', 'Diameter (km)', 'Density (kg/m^3)', 'Gravity (m/s^2)', 'Escape Velocity (km/s)', 'Rotation Period (hours)', 'Length of Day (hours)', 'Distance from Sun (10^6 km)', 'Perihelion (10^6 km)', 'Aphelion (10^6 km)', 'Orbital Period (days)', 'Orbital Velocity (km/s)', 'Orbital Inclination (deg)', 'Orbital Eccentricity', 'Obliquity to Orbit (deg)', 'Mean Temperature (C)', 'Surface Pressure (bars)', 'Number of Moons']
for col in df.columns:
    if col in columns:
        df[col] = pd.to_numeric(df[col], errors = "coerce")
```

```python
for col in df.columns:
    print(df[col].apply(type))
```

```python
# string, float, integer

# string = text, like a sequence of characters
# float = numbers that are fractional or have a decimal
# integer = whole numbers

# all the columns are still object because the columns contain mixed data types
```

```python
def attach_units(df, old_col, unit, new_col = None):
    new_column = df.copy()
    new_values = []
    for val in new_column[old_col]:
        new_values.append(val * unit)
    new_column[old_col] = new_values
    if new_col is not None:
        new_column.rename(columns = {old_col: new_col}, inplace=True)
    return new_column
```

```python
print(attach_units(df, "Mass (10^24kg)", 1e24 * u.kg))
```

```python
semi_major_axis = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2
df.insert(21, "semi-major axis (10^6 km)", semi_major_axis)
```

```python
orbital_period_days = df["Orbital Period (days)"]
new_orbital_period = attach_units(orbital_period_days, "Orbital Period (days)", u.day.to(u.year), "Orbital Period (Years)")
df.insert(11, "Orbital Period (years)", new_orbital_period)
```
