```python
# string, float, integer

# string = text, like a sequence of characters
# float = numbers that are fractional or have a decimal
# integer = whole numbers

# all the columns are still object because the columns contain mixed data types
```

```python
def attach_units(df, old_col, unit, new_col = None):
    new_df = df.copy()
    new_values = []
    for val in new_df[old_col]:
        new_values.append(val * unit)
    new_df[old_col] = new_values
    if new_col is not None:
        new_df.rename(columns = {old_col: new_col}, inplace=True)
    return new_df
```

```python
print(attach_units(df, "Mass (10^24kg)", 1e24 * u.kg))
```

```python
distance_columns = ['Diameter (km)', 'Distance from Sun (10^6 km)', 'Perihelion (10^6 km)', 'Aphelion (10^6 km)','semi-major axis (10^6 km)']  
convert_km_au = (1 * u.km).to(u.AU).value
for col_name in distance_columns:
    if col_name in df.columns:
        new_col_name = col_name + "(AU)"
        converted_values = df[col_name] * convert_km_au
        col_index = df.columns.get_loc(col_name)
        df.insert(col_index + 1, new_col_name, converted_values)
```
